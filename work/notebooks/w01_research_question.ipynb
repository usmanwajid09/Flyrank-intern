We are choosing the **Content Refresh Opportunity Scoring** lane.

### The Research Question & Business Context
* **Research Question:** *"Which high-impact pages are exhibiting search performance decay, and which of those should we prioritize for content refreshes?"*
* **Decision to Improve:** The weekly allocation of content refresh budget across the client's site.
* **Action taken:** A writer reviews the prioritized page, updates the outdated facts, improves content depth, and republishes it.
* **Cost of a wrong recommendation:**
  * **False Positive (refreshing a stable page):** Wastes production budget (e.g., $150–$300 per page) and risks disturbing a page that already ranks well.
  * **False Negative (missing a decaying page):** Leads to a silent loss of organic traffic, conversions, and organic keyword positions to competitors.
* **Why Data/ML helps:** Content decay is not a simple function of age; many old pages remain top rankers, while some new ones drop fast. Decaying patterns involve multiple signals (declines in CTR, drops in position, drops in impressions/clicks, keyword fragmentation) which are non-linear. Machine Learning can combine these noisy signals to accurately identify true performance decay.

## 3. Quick look at the data (2-3 real numbers)
Below, we load the starter dataset and compute key metrics that show why this lane is worth pursuing.


```python
import pandas as pd
import numpy as np

# Load the starter CSV
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. Base statistics
total_pages = len(df)
declining_pct = (df["trend_direction"] == "down").mean() * 100
print(f"Total pages in starter set: {total_pages}")
print(f"Percentage of pages with a 'down' trend: {declining_pct:.2f}%")

# 2. Median impressions by trend direction (High impressions mean high visibility and priority)
median_imp = df.groupby("trend_direction")["impressions_90d"].median()
print("\nMedian impressions (90d) by trend direction:")
print(median_imp.to_string())

# 3. Correlation between content age and decline (shows why content age alone is not enough)
df["is_declining"] = (df["trend_direction"] == "down").astype(int)
corr_age_decline = df["content_age_days"].corr(df["is_declining"])
print(f"\nCorrelation between content_age_days and is_declining: {corr_age_decline:.4f}")

```

### Claims and Scope: What We Can and Cannot Claim
* **What we CAN claim:** We are providing a **decision-support tool** that uses **observed** historical performance to highlight **directional** trends of content decay. It helps prioritize human review.
* **What we CANNOT claim:** We are **not** predicting Google's core algorithm, nor can we guarantee that a refresh will restore traffic (which would be a **causal proof**). Search engine result pages are dynamic, influenced by competitors, algorithm updates, and external seasonality that our model cannot observe.
