### Framing the Lane: ML Task Type
We frame this task as a **Scoring & Ranking** problem (binary classification followed by probability sorting). 

* **ML Task Type:** Supervised Binary Classification.
* **Why:** We want to estimate the probability that a specific page's search performance is declining. Predicting a raw binary class (yes/no) is not helpful because it doesn't give a priority queue. By outputting a continuous probability score (from 0.0 to 1.0) and sorting the pages by this score, we get a ranked list of content refresh opportunities.

### Success Metric
* **Core Success Metric:** **Precision@K (Precision@50)**.
* **Rationale:** A content team typically has a fixed budget to refresh $K$ pages per week (e.g., 50 pages). If our model recommends 50 pages, we want to maximize the fraction of those pages that are *actually* declining. A precision of **0.70+** (meaning 35 out of 50 recommended pages are indeed declining) is our target. This represents a ~2x to 3x improvement over the default hand-written rule baseline (~0.24 to 0.34 precision).

### Why ML Beats a Fixed Rule
A fixed rule like `stale x visible` (e.g., `days_since_last_update >= 180` and `impressions_90d >= 500`) is too crude for the following reasons:
1. **Hard Thresholds:** It treats a page updated 179 days ago as perfectly fresh, and one at 180 days as stale. It ignores pages with 499 impressions.
2. **Missing Interactions:** It cannot naturally combine multiple interacting signals (e.g., a page with average position 11.5 and a declining click-through rate, vs a page with average position 2.5 and stable impressions).
3. **No Probability Ranking:** A rule outputs a binary flag, whereas an ML model computes a smooth probability score, allowing editors to sort their queue by priority.
